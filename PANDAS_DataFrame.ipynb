```python
import numpy as np
import pandas as pd
data={
    "name":["Alice","Joe","Jack","Grace"],
    "age":[20,18,22,19],
    "marks":[50,80,30,np.nan],
    "attendence":[99,88,77,np.nan],
    "city":["delhi","mumbai","chennai","banglore"]
    }
df=pd.DataFrame(data)
print(df)
print("\n",df.head(3))  #printing first 3 rows
print("\n",df.tail(2))  #printing last 2 rows
print("\n",df.shape)    #finding rows and columns
print("\n",df.columns)  #display columns
print("\n",df.dtypes)   #find datatype
print("\n",df["city"])
print("\n",df[["age","name"]])
print("\n",df.iloc[:3])  #skip last one
print("\n",df.loc[1:3,["name","marks"]])  #include last one
print("\n",df[df["marks"]<70])
print("\n",df[df["city"]=="chennai"])
print("\n",df[(df["marks"]>70) & (df["attendence"]>80)])
print("\n",df["marks"]+5)
print("\n",df["marks"]*2)
print("\n",df["marks"].mean())
print("\n",df["marks"].sum())
print("\n",df["marks"].max())
print("\n",df["marks"].min())
print("\n",df.sort_values("marks"))
print("\n",df.sort_values("marks",ascending=False))
print("\n",df.rename(
    columns={"marks":"score",
             "attendence":"attendance_percentage"}
))
print("\n",df.isna())
print("\n",df.dropna())
print("\n",df.fillna(0))

df_for=df.copy()
df_for=df_for.ffill()
print("\n",df_for)

df_for=df.copy()
df_for=df_for.bfill()
print("\n",df_for)
```
