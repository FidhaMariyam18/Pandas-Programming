```python
#LEVEL_6

print("\tSTONE PAPER SCISSORS\t\n")
import random
choices=["stone","paper","scissors"]
while True:
  user_input=input("Enter STONE,PAPER or SCISSORS: ")
  if user_input not in choices:
    print("Invalid choice!")
    continue
  bot_choice=random.choice(choices)
  print("Bot choice is: ",bot_choice)

  if user_input==bot_choice:
    print("Match is draw!")
  elif (user_input=="stone" and bot_choice=="scissors")or\
   (user_input=="paper" and bot_choice=="stone")or\
    (user_input=="scissors" and bot_choice=="paper"):
    print("You wins")
  else:
    print("Bot wins")
  break
```

```python
#LEVEL_7

print("\tSTONE PAPER SCISSORS\t\n")
import random
choices=["stone","paper","scissors"]
while True:
  try:
    user_input=input("Enter STONE,PAPER or SCISSORS: ")
    if user_input not in choices:
      print("Invalid choice!")
      continue
    bot_choice=random.choice(choices)
    print("Bot choice is: ",bot_choice)

    if user_input==bot_choice:
      print("Match is draw!")
    elif (user_input=="stone" and bot_choice=="scissors")or\
     (user_input=="paper" and bot_choice=="stone")or\
      (user_input=="scissors" and bot_choice=="paper"):
      print("You wins")
    else:
      print("Bot wins")
  except ValueError:
    print("Error")
  play_again=input("\nDo you want to play again?(Yes/No): ")
  if play_again!="Yes":
    print("Thanks for playing the game!")
    break
```
