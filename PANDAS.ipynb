```python
#LOC(does not skip end value bcz it is a character)
import pandas as pd
data=[10,20,30,40,50]
a=pd.Series(data,index=['A','B','C','D','E'])
print(a.loc['C'])
print(a.loc['B':'D'])
```

```python
#ILOC(skip last value bcz it is a number)
data=[10,20,30]
a=pd.Series(data)
print(a.iloc[0])
print(a.iloc[:2])
```

```python
import pandas as pd
sales=pd.Series(
    [1200,850,1500,950,2000],
    index=['Monday','Tuesday','Wednesday','Thursday','Friday']
)
print(sales)
print(sales.max())
print(sales.min())
print(sales.sum())
print(sales[sales<1000])
```
