```python
import pandas as pd
l=[1,2,3,4,5]
s=pd.Series(l)
print(s)
```

```python
l=[1,2,3]
s=pd.Series(l,index=['A','B','C'])
print(s)
```

```python
data={'name':'rih','class':'BCA','age':18}
s=pd.Series(data)
print(s)
```

```python
#LOC(does not skip end value bcz it is a character)
data=[10,20,30,40,50]
a=pd.Series(data,index=['A','B','C','D','E'])
print(a.loc['C'])
print(a.loc['B':'D'])
```

```python
#ILOC(skip last value bcz it is a number)
data=[10,20,30]
a=pd.Series(data)
print(a.iloc[0])
print(a.iloc[:2])
```
